## Visual 5

Visual 5 will be a map of the different police districts of Chicago, highlighting social and/or economically poor districts in combination with Crime rates in that specific district. This will give insight on the question: does poverty have a big influence on crime?

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.offline as pyo
from IPython.display import HTML

# 📄 Load CSV files
crime_df = pd.read_csv('dataset/graph4.csv')             # crime data
trust_safety_df = pd.read_csv('dataset/sentfinal.csv')   # socioeconomic and safety data

# Clean and merge
crime_counts = crime_df['District'].value_counts().reset_index()
crime_counts.columns = ['District', 'Total_Crimes']
trust_safety_df = trust_safety_df.rename(columns={'DISTRICT': 'District'})

merged_df = pd.merge(trust_safety_df, crime_counts, on='District', how='inner')

# Extract list of unique years
years = sorted(merged_df['YEAR'].unique())

# Create figure
fig = go.Figure()

for year in years:
    year_df = merged_df[merged_df['YEAR'] == year].sort_values(by='SAFETY', ascending=False).reset_index(drop=True)
    x_labels = year_df['District'].astype(str)

    custom_data = year_df[['District']]

    fig.add_trace(go.Bar(
        name='Low Income',
        x=x_labels,
        y=year_df['S_INCOME_LOW'],
        visible=(year == years[0]),
        customdata=custom_data,
        hovertemplate='Low Income: %{y:.1f}%<extra></extra>'
    ))

    fig.add_trace(go.Bar(
        name='Medium Income',
        x=x_labels,
        y=year_df['S_INCOME_MEDIUM'],
        visible=(year == years[0]),
        customdata=custom_data,
        hovertemplate='Medium Income: %{y:.1f}%<extra></extra>'
    ))

    fig.add_trace(go.Bar(
        name='High Income',
        x=x_labels,
        y=year_df['S_INCOME_HIGH'],
        visible=(year == years[0]),
        customdata=custom_data,
        hovertemplate='High Income: %{y:.1f}%<extra></extra>'
    ))

    fig.add_trace(go.Bar(
        name='Safety Score',
        x=x_labels,
        y=year_df['SAFETY'],
        visible=(year == years[0]),
        customdata=custom_data,
        hovertemplate='Safety Score: %{y:.1f}%<extra></extra>'
    ))

# Create dropdown buttons
buttons = []
for i, year in enumerate(years):
    visible = [False] * len(years) * 4
    for j in range(4):
        visible[i * 4 + j] = True

    year_df = merged_df[merged_df['YEAR'] == year].sort_values(by='SAFETY', ascending=False).reset_index(drop=True)
    x_labels = year_df['District'].astype(str)

    buttons.append(dict(
        label=str(year),
        method='update',
        args=[
            {'visible': visible},
            {
                'layout': {
                    'title': f'Income Levels and Safety Score by District - {year}',
                    'xaxis': {
                        'tickmode': 'array',
                        'tickvals': x_labels,
                        'ticktext': x_labels,
                        'tickangle': -45,
                        'title': 'District (High Safety → Low Safety)'
                    }
                }
            }
        ]
    ))

# Initial year axis setup
initial_year = years[0]
initial_year_df = merged_df[merged_df['YEAR'] == initial_year].sort_values(by='SAFETY', ascending=False).reset_index(drop=True)
initial_x_labels = initial_year_df['District'].astype(str)

# Update layout
fig.update_layout(
    title=f'Income Levels and Safety Score by District',
    barmode='group',
    legend=dict(
        orientation="h",
        y=1.1,
        x=0,
        xanchor='left',
        yanchor='bottom',
        font=dict(size=12)
    ),
    updatemenus=[{
        'buttons': buttons,
        'direction': 'down',
        'showactive': True,
        'x': 0.60,
        'xanchor': 'left',
        'y': 1.18,
        'yanchor': 'top'
    }],
    margin=dict(t=150),
    xaxis=dict(
        tickmode='array',
        tickvals=initial_x_labels,
        ticktext=initial_x_labels,
        tickangle=-45,
        title='District (High Safety → Low Safety)'
    ),
    yaxis_title='Percentage',
    hovermode='x unified',
    hoverlabel=dict(
        bgcolor="white",
        font_size=14,
        font_family="Arial"
    ),
    height=600,
)

# Render the chart
plot_html = pyo.plot(fig, include_plotlyjs='cdn', output_type='div')
HTML(plot_html)
```

## Visual 6

Visual 6 will be a map of the different police districts of chicago, highlighting the sediment score in combination with crime rates in that specific district. This will give insight on the question: Does a lower crime-rate result in a higher sediment score?

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.offline as pyo
from IPython.display import HTML

# Load data
crime_df = pd.read_csv('dataset/graph4.csv')
trust_df = pd.read_csv('dataset/sentfinal.csv')

# Prep and rename
crime_counts = crime_df['District'].value_counts().reset_index()
crime_counts.columns = ['District', 'Crime_Count']
trust_df = trust_df.rename(columns={'DISTRICT': 'District'})

# Merge on district
merged_df = pd.merge(crime_counts, trust_df[['District', 'TRUST', 'YEAR']], on='District', how='inner')

# Get unique years
years = sorted(merged_df['YEAR'].unique())

# Initialize figure
fig = go.Figure()

for year in years:
    df_year = merged_df[merged_df['YEAR'] == year].sort_values(by='Crime_Count', ascending=False).reset_index(drop=True)
    
    x_pos = list(range(len(df_year)))
    district_labels = df_year['District']
    customdata = df_year[['District', 'Crime_Count', 'TRUST']].values

    hovertemplate = (
        "<b>District: %{customdata[0]}</b><br>" +
        "Crime Count: %{customdata[1]}<br>" +
        "Trust: %{customdata[2]:.1f}%<extra></extra>"
    )

    # Crime Count
    fig.add_trace(go.Bar(
        x=x_pos,
        y=df_year['Crime_Count'],
        name='Crime Count',
        marker_color='indianred',
        yaxis='y1',
        visible=(year == years[0]),
        customdata=customdata,
        hovertemplate=hovertemplate
    ))

    # Trust
    fig.add_trace(go.Bar(
        x=x_pos,
        y=df_year['TRUST'],
        name='Trust (%)',
        marker_color='steelblue',
        yaxis='y2',
        visible=(year == years[0]),
        customdata=customdata,
        hovertemplate=hovertemplate
    ))

# Dropdown menu
buttons = []
for i, year in enumerate(years):
    visibility = [False] * len(years) * 2
    visibility[i * 2] = True     # Crime
    visibility[i * 2 + 1] = True # Trust

    df_year = merged_df[merged_df['YEAR'] == year].sort_values(by='Crime_Count', ascending=False).reset_index(drop=True)
    x_ticks = list(range(len(df_year)))
    tick_text = df_year['District'].tolist()

    buttons.append(dict(
        label=str(year),
        method='update',
        args=[
            {'visible': visibility},
            {
                'layout': {
                    'title': f'Crime Count and Trust Value by District (Sorted by Crime Count) - {year}',
                    'xaxis': {
                        'tickmode': 'array',
                        'tickvals': x_ticks,
                        'ticktext': tick_text,
                        'tickangle': -45,
                        'title': 'District (High Crime → Low Crime)'
                    }
                }
            }
        ]
    ))

# Initial year setup
initial_year = years[0]
initial_df = merged_df[merged_df['YEAR'] == initial_year].sort_values(by='Crime_Count', ascending=False).reset_index(drop=True)

# Layout
fig.update_layout(
    title=f'Crime Count and Trust Value by District (Sorted by Crime Count)',
    xaxis=dict(
        tickmode='array',
        tickvals=list(range(len(initial_df))),
        ticktext=initial_df['District'],
        tickangle=-45,
        title='District (High Crime → Low Crime)'
    ),
    yaxis=dict(
        title='Crime Count',
        showgrid=False,
        zeroline=False,
    ),
    yaxis2=dict(
        title='Trust (%)',
        overlaying='y',
        side='right',
        range=[0, 100],
        showgrid=False,
        zeroline=False,
    ),
    barmode='group',
    legend=dict(
        orientation="h",
        yanchor="top",
        y=0.95,
        xanchor="center",
        x=0.5,
        font=dict(size=12)
    ),
    margin=dict(t=110),
    updatemenus=[{
        'buttons': buttons,
        'direction': 'down',
        'showactive': True,
        'x': 0.01,
        'xanchor': 'left',
        'y': 1.15,
        'yanchor': 'top'
    }]
)

# Render output
plot_html = pyo.plot(fig, include_plotlyjs='cdn', output_type='div')
HTML(plot_html)
```

```python
# LET OP!
# omdat geopandas niet werkt via github pages, heb ik via de .gitignore ingevoerd dat github nooit de checkpoints van de .ipynb mag verwijderen. 
# hierdoor kan hij de map altijd laten zien zonder dat je de error krijgt dat geopandas niet beschikbaar is. 
# omlokaal de map te creeren kun je de ''' weghalen boven en onderaan de code en het runnen. 



'''
import pandas as pd
# import geopandas as gpd
import folium
from folium.plugins import MarkerCluster
import branca.colormap as cm
from IPython.display import IFrame

from plotly.offline import init_notebook_mode
init_notebook_mode(connected=True)

crime_df = pd.read_csv('dataset/graph4.csv')

districts_gdf = gpd.read_file('map.geojson')

if crime_df.empty:
    print("Error: Crime dataset is empty. Please check the URL or file path.")
    exit()

print("Unique District values in crime dataset (before cleaning):", sorted(crime_df['District'].astype(str).unique()))


crime_df['District'] = crime_df['District'].astype(str).str.replace('.0', '').str.strip()
crime_df = crime_df[crime_df['District'] != 'nan']

print("Unique District values in crime dataset (after cleaning):", sorted(crime_df['District'].unique()))

crime_counts = crime_df.groupby('District').size().reset_index(name='Crime_Count')
print("Crime counts by district:\n", crime_counts)

districts_gdf = gpd.read_file('map.geojson')

print("Unique dist_num values in GeoJSON:", sorted(districts_gdf['dist_num'].astype(str).unique()))

districts_gdf['dist_num'] = districts_gdf['dist_num'].astype(str).str.strip()

districts_gdf = districts_gdf.merge(crime_counts, left_on='dist_num', right_on='District', how='left')
districts_gdf['Crime_Count'] = districts_gdf['Crime_Count'].fillna(0)

print("Merged GeoDataFrame:\n", districts_gdf[['dist_num', 'dist_label', 'Crime_Count']])

chicago_map = folium.Map(location=[41.8781, -87.6298], zoom_start=11, tiles='CartoDB positron')

colormap = cm.LinearColormap(
    colors=['#f0f9e8', '#bae4bc', '#7bccc4', '#43a2ca', '#0868ac'],
    vmin=districts_gdf['Crime_Count'].min(),
    vmax=districts_gdf['Crime_Count'].max() if districts_gdf['Crime_Count'].max() > 0 else 1,
    caption='Crime Count by Police District'
)

folium.GeoJson(
    districts_gdf,
    style_function=lambda feature: {
        'fillColor': colormap(feature['properties']['Crime_Count']),
        'color': 'black',
        'weight': 1,
        'fillOpacity': 0.7,
    },
    tooltip=folium.GeoJsonTooltip(
        fields=['dist_label', 'Crime_Count'],
        aliases=['District', 'Crime Count'],
        localize=True
    )
).add_to(chicago_map)

colormap.add_to(chicago_map)

chicago_map.save('map.html')
'''
```
