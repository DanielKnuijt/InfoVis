## Visual 1
Theft, Battery and Criminal Damage were the top 3 most common crimes committed in Chicago. This graph tells us the percentage of one of top 3 crimes, compared to the total crimes committed in a specific year. Every year, except for 2020, Theft was the most popular crime committed. We believe that in 2020, less Theft was committed because of the COVID-19 pandemic

This graph is based on the aggregated dataset containing only data from 2018-2023. The most common crime types are counted where empty rows are ignored. At last, this result is presented as percentages of the total amount of crimes committed during 6-year time span.

## Visual 3
Crimes can be committed in different places in the city. This graph shows where crimes are committed most often between 2018 and 2023. According to the graph, the most popular places are the street, in someone's apartment and then someone's residence. Something we noticed is the lack of crimes committed in alleyways, where normally you'd expect a lot more crimes according to Hollywood.

This graph is based on the aggregated dataset containing only data from 2018-2023. The most common crime-scenes are counted and put into a pie-chart. This pie chart clearly shows the most popular crime-scenes, together with some less-popular crime-scenes. This pie-chart shows the average of the years 2018-2023. Finally the Pie-chart is transformed into a donut-chart for aestetic purposes.

```python
'''
import pandas as pd
import plotly.express as px
import plotly.offline as pyo
from IPython.display import HTML

# Load and filter the dataset
df = pd.read_csv('dataset/graph123.csv')

# Filter by year and drop unused columns
df = df[(df['Year'] >= 2018) & (df['Year'] <= 2023)]
df = df.drop(columns=['ID', 'Case Number'], errors='ignore')

# Drop missing values and convert 'Location Description' to string
df = df.dropna(subset=['Location Description'])
df['Location Description'] = df['Location Description'].str.title()

# Count occurrences of each location
location_counts = df['Location Description'].value_counts().reset_index()
location_counts.columns = ['Location Description', 'Count']

# Optional: show only top N locations to reduce clutter
top_n = 10
location_counts = location_counts.head(top_n)

# Create donut chart
fig = px.pie(
    location_counts,
    values='Count',
    names='Location Description',
    title='Top Locations of Crime (2018–2023)',
    hole=0.6,
    color_discrete_sequence=[
        "#636EFA", "#EF553B", "#00CC96", "#AB63FA", "#FFA15A",
        "#19D3F3", "#FF6692", "#B6E880", "#FF97FF", "#FECB52"
    ]
)

fig.update_traces(
    textposition='inside',
    textfont_color='white',
    insidetextorientation='auto'
)

# Improve layout
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(
    showlegend=True,
    legend_title_text='Location',
    title_font=dict(size=22),
    margin=dict(t=60, b=40, l=20, r=20)
)

# Render HTML using plotly.offline
plot_html = pyo.plot(fig, include_plotlyjs='cdn', output_type='div')

# Display HTML in notebook (safe for MyST/Jupyter Book)
HTML(plot_html)
'''

import pandas as pd
import plotly.express as px
import plotly.offline as pyo
from IPython.display import HTML

# Load and filter the dataset
df = pd.read_csv('dataset/graph123.csv')

# Filter by year and drop unused columns
df = df[(df['Year'] >= 2018) & (df['Year'] <= 2023)]
df = df.drop(columns=['ID', 'Case Number'], errors='ignore')

# Drop missing values and convert 'Location Description' to string
df = df.dropna(subset=['Location Description'])
df['Location Description'] = df['Location Description'].str.title()

# Count occurrences of each location
location_counts = df['Location Description'].value_counts().reset_index()
location_counts.columns = ['Location Description', 'Count']

# Optional: show only top N locations to reduce clutter
top_n = 10
location_counts = location_counts.head(top_n)

# Identify top 4 sectors for text labels
top4_sectors = location_counts.nlargest(4, 'Count')['Location Description'].tolist()

# Prepare custom text: label + percent for top 4, else blank
total = location_counts['Count'].sum()
location_counts['Percent'] = (location_counts['Count'] / total) * 100

def custom_label(row):
    if row['Location Description'] in top4_sectors:
        return f"{row['Location Description']}<br>{row['Percent']:.1f}%"
    else:
        return ""

location_counts['CustomText'] = location_counts.apply(custom_label, axis=1)

# Create donut chart
fig = px.pie(
    location_counts,
    values='Count',
    names='Location Description',
    title='Top Locations of Crime (2018–2023)',
    hole=0.6,
    color_discrete_sequence=[
        "#636EFA", "#EF553B", "#00CC96", "#AB63FA", "#FFA15A",
        "#19D3F3", "#FF6692", "#B6E880", "#FF97FF", "#FECB52"
    ]
)

# Update traces: use custom text, show only those labels
fig.update_traces(
    text=location_counts['CustomText'],
    textposition='inside',
    textfont_color='white',
    textinfo='text',
    insidetextorientation='auto',
    hovertemplate='%{percent:.2%}<extra></extra>'
)


# Improve layout
fig.update_layout(
    showlegend=True,
    legend_title_text='Location',
    title_font=dict(size=22),
    margin=dict(t=60, b=40, l=20, r=20),
    hoverlabel=dict(
        font_color='white',
        bgcolor='rgba(0,0,0,0.7)',
        bordercolor='white'
    )
)

# Render HTML using plotly.offline
plot_html = pyo.plot(fig, include_plotlyjs='cdn', output_type='div')

# Display HTML in notebook (safe for MyST/Jupyter Book)
HTML(plot_html)


```

## Visual 5

Visual 5 will be a map of the different police districts of Chicago, highlighting social and/or economically poor districts in combination with Crime rates in that specific district. This will give insight on the question: does poverty have a big influence on crime?

```python
import pandas as pd
import plotly.graph_objects as go

# 📄 Load CSV files
crime_df = pd.read_csv('dataset/graph4.csv')             # crime data
trust_safety_df = pd.read_csv('dataset/sentfinal.csv')   # socioeconomic and safety data

# Clean and merge
crime_counts = crime_df['District'].value_counts().reset_index()
crime_counts.columns = ['District', 'Total_Crimes']
trust_safety_df = trust_safety_df.rename(columns={'DISTRICT': 'District'})

merged_df = pd.merge(trust_safety_df, crime_counts, on='District', how='inner')

# Extract list of unique years
years = sorted(merged_df['YEAR'].unique())

# Create a trace set for each year
fig = go.Figure()

for year in years:
    year_df = merged_df[merged_df['YEAR'] == year].sort_values(by='SAFETY', ascending=False).reset_index(drop=True)
    x_pos = list(range(len(year_df)))

    fig.add_trace(go.Bar(
        name='Low Income',
        x=x_pos,
        y=year_df['S_INCOME_LOW'],
        visible=(year == years[0]),
        hovertemplate='<b>%{text}</b><br>Low Income: %{y}',
        # text=year_df['District']
    ))

    fig.add_trace(go.Bar(
        name='Medium Income',
        x=x_pos,
        y=year_df['S_INCOME_MEDIUM'],
        visible=(year == years[0]),
        hovertemplate='<b>%{text}</b><br>Medium Income: %{y}',
        # text=year_df['District']
    ))

    fig.add_trace(go.Bar(
        name='High Income',
        x=x_pos,
        y=year_df['S_INCOME_HIGH'],
        visible=(year == years[0]),
        hovertemplate='<b>%{text}</b><br>High Income: %{y}',
        # text=year_df['District']
    ))

    fig.add_trace(go.Bar(
        name='Safety Score',
        x=x_pos,
        y=year_df['SAFETY'],
        visible=(year == years[0]),
        hovertemplate='<b>%{text}</b><br>Safety Score: %{y}',
        # text=year_df['District']
    ))

# Create dropdown buttons
buttons = []
for i, year in enumerate(years):
    visible = [False] * len(years) * 4
    for j in range(4):
        visible[i * 4 + j] = True
    buttons.append(dict(
        label=str(year),
        method='update',
        args=[{'visible': visible},
              {'title': f'Income Levels and Safety Score by District - {year}'}]
    ))

# Update layout
fig.update_layout(
    updatemenus=[{
        'buttons': buttons,
        'direction': 'down',
        'showactive': True,
        'x': 1.15,
        'xanchor': 'left',
        'y': 1.1,
        'yanchor': 'top'
    }],
    barmode='group',
    xaxis=dict(
        tickmode='array',
        tickvals=list(range(len(merged_df['District'].unique()))),
        ticktext=merged_df['District'].unique(),
        tickangle=-45,
        title='District (High Safety → Low Safety)'
    ),
    yaxis_title='Value',
    title=f'Income Levels and Safety Score by District - {years[0]}',
    hovermode='x unified'
)

fig.show()
```

## Visual 6

Visual 6 will be a map of the different police districts of chicago, highlighting the sediment score in combination with crime rates in that specific district. This will give insight on the question: Does a lower crime-rate result in a higher sediment score?

```python
import pandas as pd
import plotly.graph_objects as go

# Load data (replace with your paths)
crime_df = pd.read_csv('dataset/graph4.csv')
trust_df = pd.read_csv('dataset/sentfinal.csv')

# Count crimes per district
crime_counts = crime_df['District'].value_counts().reset_index()
crime_counts.columns = ['District', 'Crime_Count']

# Prepare trust_df
trust_df = trust_df.rename(columns={'DISTRICT': 'District'})

# Merge on district
merged_df = pd.merge(crime_counts, trust_df[['District', 'TRUST']], on='District', how='inner')

# Sort descending by crime count
merged_df = merged_df.sort_values(by='Crime_Count', ascending=False).reset_index(drop=True)

# Numeric x-axis positions
x_pos = list(range(len(merged_df)))

fig = go.Figure()

# Crime count bar on primary y-axis (left)
fig.add_trace(go.Bar(
    x=x_pos,
    y=merged_df['Crime_Count'],
    name='Crime Count',
    marker_color='indianred',
    yaxis='y1'
))

# Trust bar on secondary y-axis (right)
fig.add_trace(go.Bar(
    x=x_pos,
    y=merged_df['TRUST'],
    name='Trust (%)',
    marker_color='steelblue',
    yaxis='y2'
))

fig.update_layout(
    title='Crime Count and Trust Value by District (Sorted by Crime Count)',
    xaxis=dict(
        tickmode='array',
        tickvals=x_pos,
        ticktext=merged_df['District'],
        tickangle=-45,
        title='District (High Crime → Low Crime)'
    ),
    yaxis=dict(
        title='Crime Count',
        showgrid=False,
        zeroline=False,
    ),
    yaxis2=dict(
        title='Trust (%)',
        overlaying='y',
        side='right',
        range=[0, 100],  # Trust is percent, fix range 0-100
        showgrid=False,
        zeroline=False,
    ),
    barmode='group',
    legend=dict(x=0.85, y=1.25),
    margin=dict(t=100)
)

fig.show()
```
